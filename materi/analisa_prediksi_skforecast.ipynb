## 2. Bagaimana bentuk data trainingnya?

Pada kasus ini, data training dibentuk menggunakan metode supervised learning. Model menggunakan data historis permintaan listrik (Demand) dan variabel eksternal berupa Temperature untuk memprediksi nilai Demand berikutnya.

### Input (X)
Input yang digunakan terdiri dari:
- lag_1 : nilai Demand 1 periode sebelumnya
- lag_2 : nilai Demand 2 periode sebelumnya
- lag_3 : nilai Demand 3 periode sebelumnya

- lag_5 : nilai Demand 5 periode sebelumnya

- lag_7 : nilai Demand 7 periode sebelumnya

### Output (y)

Jadi, bentuk data training pada kasus ini adalah:
- Input (X) = lag_1 sampai lag_7 dan Temperature.
- Output (y) = Demand.


## 3. Apa itu Lag ?

Lag adalah nilai masa lalu dari varibale target yang digunakan sebagai fitur prediksi

pada kasus ini

* lag_1 = Demand 1 hari sebelumnya
* lag_2 = Demand 2 hari sebelumnya
* lag_3 = nilai Demand 3 periode sebelumnya,
* lag_4 = nilai Demand 4 periode sebelumnya,
* lag_5 = nilai Demand 5 periode sebelumnya,
* lag_6 = nilai Demand 6 periode sebelumnya,
* lag_7 = nilai Demand 7 periode sebelumnya.

Ketujuh nilai tersebut digunakan oleh model untuk memprediksi nilai Demand pada periode selanjutnya.

## 4. Jelaskan proses analisis yang dilakukan dari kasus di atas

### a. Mengambil dataset

Dataset `vic_electricity` digunakan sebagai sumber data permintaan listrik.

```python
data = fetch_dataset(name='vic_electricity')
```

---

### b. Membagi data menjadi training dan testing

Data dipisahkan berdasarkan urutan waktu agar tidak terjadi kebocoran data.

```python
data_train = data.loc[:'2014-12-21']
data_test = data.loc['2014-12-22':]
```

---

### c. Membuat model forecasting

Model dibuat menggunakan `ForecasterRecursive` dengan algoritma `LGBMRegressor` dan menggunakan 7 lag.

```python
forecaster = ForecasterRecursive(
    regressor=LGBMRegressor(random_state=123),
    lags=7
)
```

---

### d. Melatih model

Model dilatih menggunakan data Demand sebagai target dan Temperature sebagai variabel eksternal.

```python
forecaster.fit(
    y=data_train['Demand'],
    exog=data_train['Temperature']
)
```

---

### e. Melakukan analisis feature importance

Feature importance digunakan untuk mengetahui fitur yang paling berpengaruh terhadap prediksi.

```python
forecaster.get_feature_importances()
```

---

### f. Analisis menggunakan SHAP

SHAP digunakan untuk menjelaskan kontribusi setiap fitur terhadap hasil prediksi.

```python
explainer = shap.TreeExplainer(forecaster.regressor)
```

---

### g. Melakukan prediksi

Model digunakan untuk memprediksi permintaan listrik pada beberapa periode ke depan.

```python
predictions = forecaster.predict(
    steps=10,
    exog=data_test['Temperature']
)
```

Berdasarkan proses tersebut, model dapat memprediksi permintaan listrik sekaligus memberikan penjelasan mengenai faktor-faktor yang memengaruhi hasil prediksi.
