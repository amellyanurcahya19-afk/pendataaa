# Pengantar Data Mining

**Pengertian Data Mining**

Data mining adalah proses menggali dan menemukan pola, hubungan, atau informasi penting dari kumpulan data yang besar menggunakan teknik statistik, matematika, dan komputasi. Intinya, data mining mengubah data mentah menjadi informasi yang bermakna dan bisa digunakan untuk pengambilan keputusan.

**Tujuan Data Mining**

Tujuan utana data mining anatar lain, yaitu:

1. Menemukan pola tersembunyi dalam data yang tidak terlihat secara langsung
2. Membantu pengambilan keputusan berdasarkan data,bukan hanya perkiraan
3. Kemampuan Data mining untuk mencari informasi bisnis yang berharga dari basis data yang sangat besar
4. Mengelompokkan data berdasarkan karakteristik tertentu

**Manfaat Data Mining**

1. Meningkatkan efisiensi kerja karena analisis dilakukan secara otomatis
2. Membantu perencanaan dan strategi jangka panjang
3. Mendeteksi masalah lebih dini,seperti penurunan performa atau anomali data
4. Menemukan pola dan informasi tersembunyi

#  Explorasi iris menggunakan collab (python)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
df = pd.DataFrame(
    iris.data,
    columns=['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
)
df['species'] = iris.target
df['species'] = df['species'].map({0: 'Iris-setosa', 1: 'Iris-versicolor', 2: 'Iris-virginica'})
```

```python
print("Contoh data:")
display(df.head())
```

```python
print("\nInformasi dataset:")
df.info()
```

```python
print("\nStatistik deskriptif:")
display(df.describe())
```

```python
print("\nJumlah data kosong:")
print(df.isnull().sum())
```

```python
plt.figure()
sns.scatterplot(
    data=df,
    x='sepal_length',
    y='sepal_width',
    hue='species'
)
plt.title('Scatter Plot Sepal Length vs Sepal Width')
plt.show()
```

```python
plt.figure()
sns.boxplot(
    data=df,
    x='species',
    y='petal_length'
)
plt.title('Box Plot Petal Length')
plt.show()
```

```python
# Ambil kolom numerik saja
df_numeric = df.select_dtypes(include=["int64", "float64"])

for col in df_numeric.columns:
    print("=== Statistik Deskriptif untuk", col, "===")
    print("Jumlah data       :", df_numeric[col].count())
    print("Rata-rata         :", round(df_numeric[col].mean(), 6))
    print("Nilai minimum     :", df_numeric[col].min())
    print("Q1                :", df_numeric[col].quantile(0.25))
    print("Median (Q2)       :", df_numeric[col].median())
    print("Q3                :", df_numeric[col].quantile(0.75))
    print("Nilai maksimum    :", df_numeric[col].max())
    print("Skewness          :", round(df_numeric[col].skew(), 6))

    # Modus versi aman
    modus = df_numeric[col].mode()
    if len(modus) > 0:
        print("Modus             :", modus.iloc[0])
    else:
        print("Modus             : Tidak ada")

    print("Standar deviasi   :", round(df_numeric[col].std(), 6))
    print("Variansi          :", round(df_numeric[col].var(), 6))
    print()
```
